```python
import requests
import pandas as pd
import psycopg2

base_url = 'https://www.anapioficeandfire.com/api/'
```

```python
book_list = []
page = 1

while True:
    book_list_response = requests.get(f'{base_url}books', params={'page': page, 'pageSize': 50}).json()
    book_list.extend(book_list_response)
    if not book_list_response:
        break
    page += 1

book_df = pd.DataFrame(book_list)
book_df
```

```python
houses_list = []
page = 1
while True:
    houses_responce = requests.get(f'{base_url}houses', params={'page': page, 'pageSize': 50}).json()
    houses_list.extend(houses_responce)
    if not houses_responce:
        break
    page +=1
    
houses_df = pd.DataFrame(houses_list)
print(len(houses_df))
houses_df.head(5)
```

```python
houses_with_titles_list = []
page = 1
while True:
    houses_with_titles_responce = requests.get(f'{base_url}houses', params={'hasTitles': True, 'page': page, 'pageSize': 50}).json()
    houses_with_titles_list.extend(houses_with_titles_responce)
    if not houses_with_titles_responce:
        break
    page +=1
    
houses_with_titles_df = pd.DataFrame(houses_with_titles_list)
print(len(houses_with_titles_df))
houses_with_titles_df.head()
```
